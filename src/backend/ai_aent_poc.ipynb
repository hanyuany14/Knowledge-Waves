```python
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
import requests

# 設定範圍和 API URL
SCOPES = ["https://www.googleapis.com/auth/datastudio.readonly"]
API_URL = "https://datastudio.googleapis.com/v1/assets:search"


# 初始化 OAuth 2.0 驗證
def authenticate():
    # 使用 credentials.json 初始化 OAuth 流程
    flow = InstalledAppFlow.from_client_secrets_file(CREDENTIALS_PATH, SCOPES)
    creds = flow.run_local_server(port=0)
    return creds


# 查詢資產 (Looker Studio 報表/圖表)
def search_assets(creds, asset_type="REPORT", title=None, owner=None, include_trashed=False):
    headers = {"Authorization": f"Bearer {creds.token}"}
    params = {
        "assetTypes": asset_type,
        "includeTrashed": str(include_trashed).lower(),  # 'true' or 'false'
        "pageSize": 10,  # 調整返回的結果數量
    }
    if title:
        params["title"] = title
    if owner:
        params["owner"] = owner

    # 發送 GET 請求
    response = requests.get(API_URL, headers=headers, params=params)

    if response.status_code == 200:
        assets = response.json().get("assets", [])
        for asset in assets:
            print(f"asset: {asset}")
        return assets
    else:
        print("Error:", response.status_code, response.text)
        return []


if __name__ == "__main__":
    # 認證
    credentials = authenticate()

    # 查詢 Looker Studio 資產
    print("查詢所有報表：")
    search_assets(credentials, asset_type="REPORT")

    # print("\n查詢特定圖表：")
    # search_assets(credentials, asset_type="CHART", title="HW3 - Looker Practice")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from PIL import Image
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# # 設定 Chrome 瀏覽器選項

# chrome_options = Options()
# chrome_options.add_argument("--headless")  # 無頭模式
# chrome_options.add_argument("--window-size=1920x1080")

# # 指定 ChromeDriver 的路徑
# service = Service("/usr/local/bin/chromedriver")  # 替換為您的 ChromeDriver 路徑

# # 初始化 WebDriver
# driver = webdriver.Chrome(service=service, options=chrome_options)

# 自動下載並安裝匹配的 ChromeDriver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

try:
    # 加載目標頁面
    driver.get(
        "https://lookerstudio.google.com/u/0/reporting/80e76648-4635-4d46-a953-8e466ab13235/page/p_vgrjccaxmd/edit"
    )

    # 獲取所有 iframe
    iframes = driver.find_elements(By.TAG_NAME, "iframe")
    print(f"找到 {len(iframes)} 個 iframe")

    for index, iframe in enumerate(iframes):
        try:
            # 切換到當前 iframe
            driver.switch_to.frame(iframe)
            print(f"正在檢查 iframe {index + 1}")

            # 嘗試查找 <svg> 元素
            svg_element = WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, "//svg")))
            print(f"在 iframe {index + 1} 找到 SVG 元素")

            # 獲取 SVG 的 HTML
            svg_html = svg_element.get_attribute("outerHTML")

            # 保存 SVG 文件
            with open(f"chart_{index + 1}.svg", "w", encoding="utf-8") as file:
                file.write(svg_html)
                print(f"SVG 圖表已保存為 chart_{index + 1}.svg")

            break  # 成功找到後結束迴圈
        except Exception as e:
            print(f"在 iframe {index + 1} 未找到 SVG 元素: {e}")
        finally:
            # 切換回主頁面
            driver.switch_to.default_content()

finally:
    driver.quit()
```
